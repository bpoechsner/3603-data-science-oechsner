# 📘 01 - Scalar Types

Review and practice working with Python's basic scalar data types.

## ✅ Learning Goals
- Recognize scalar types (`int`, `float`, `str`, `bool`)
- Use the `type()` function to identify types, and convert between types with casting
- Perform arithmetic operations and understand what each operator does
- Use variables, assignment, and comparison operators

## 🔧 Define Some Variables

````python
num = 42
pi = 3.14159
name = 'Python'
is_cool = True
````

## 🔍 Check Their Types

````python
print(type(num))
print(type(pi))
print(type(name))
print(type(is_cool))
````

## ➕ Arithmetic Operators
Python supports the usual math operators, plus a couple that might be new to you: `//` (floor division, drops the remainder) and `**` (exponent).

````python
x = 10
y = 3
print('Add:', x + y)
print('Subtract:', x - y)
print('Multiply:', x * y)
print('Divide:', x / y)
print('Floor Divide:', x // y)
print('Remainder (Modulo):', x % y)
print('Exponent:', x ** 2)
````

## 🔄 Type Casting
Sometimes you need to convert a value from one type to another — for example, `input()` always returns a `str`, even if the user types a number.

````python
age_text = '21'
age_number = int(age_text)
print(age_number + 1)
print(type(age_number))
````

### Practice 3 — Modulo and Floor Division (harder)
Given `total_minutes = 137`, use `//` and `%` to compute the number of whole hours and the leftover minutes, then print something like `2 hours and 17 minutes`.

````python
total_minutes = 137
whole_hours = total_minutes // 60
leftover_minutes = total_minutes % 60
print(f"{whole_hours} hours and {leftover_minutes} minutes")
````

## 🔥 Challenge (Optional) — The Floating-Point Gotcha
Try running this:
```python
print(0.1 + 0.2)
```
Not `0.3`? That's not a bug — computers store floats in binary, and most decimal fractions can't be represented exactly, so tiny rounding errors creep in. This trips up real data science code all the time (comparing floats with `==`).

Try it: write a function `close_enough(a, b)` that returns `True` if two floats are equal to within `0.0001`, instead of using `==` directly. (Hint: compare `abs(a - b)` to a small number.)

````python
def close_enough(a, b):
    result = abs(a - b) < 0.0001
    return result
    pass

print(close_enough(0.1 + 0.2, 0.3))
````
